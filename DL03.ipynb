```python
import tensorflow as tf
x_data = tf.constant([[2.],[4.],[6.0],[8.],[10.]],dtype=tf.float64)
y_data = tf.constant([[0.],[0.],[0.0],[1.],[0.]],dtype=tf.float64)
```

```python
W = tf.Variable(tf.random.normal([1,1],dtype=tf.float64, name="weight") )
b = tf.Variable(tf.random.normal([1], dtype=tf.float64, name="bias"))
```

```python
optimizer = tf.keras.optimizers.SGD(learning_rate=0.01)
```

라텍(LaTex) : 수학식, 논문, 보고서 문서
$$
  \sigma(x) = \frac{1}{1+e^{-x}}
$$
$$
  Loss = -\frac{1}{N}\sum\left[y\log(\hat{y})
  +(1-y)\log(1-\hat{y})\right]
$$

```python
from numpy import gradient
for step in range(10001):
  with tf.GradientTape() as tape:
    pred = tf.sigmoid(tf.matmul(x_data, W)+ b)
    # 이진 교차 엔트로피 (Binary Cross Entropy : BOE 손실함수)
    loss = -tf.reduce_mean(y_data * tf.math.log(pred) + (1-y_data)*tf.math.log(1-pred))
  gradient = tape.gradient(loss,[W,b])
  optimizer.apply_gradients(zip(gradient,[W,b]))

  if step% 2000 == 0:
    print(f"{step} 번째 학습중 ,loss = {loss :.3f}")
```

```python
test_data = [[7.0],[4.0]]
test_data = tf.constant(test_data, dtype=tf.float64)

final_pred = tf.sigmoid(tf.matmul(test_data, W)+ b)
precated = tf.cast(final_pred > 0.5, dtype=tf.float64)
precated
```
